## 2. ¿Qué es la población en este conjunto de datos?

Recordemos la definición: la **población** es el conjunto de *todos* los elementos sobre los que queremos hacer una afirmación o análisis.

Aquí tenemos una fila por cada departamento de Colombia, más Bogotá D.C. — es decir, **todas** las entidades territoriales del país están incluidas. Esto es distinto a una encuesta como la Gran Encuesta Integrada de Hogares (GEIH), que solo entrevista a una parte de los hogares.

**Pregunta para discutir en clase:** si nuestro interés es "comparar la competitividad entre los departamentos de Colombia", ¿este conjunto de datos es una población o una muestra? ¿Por qué?

```r
nrow(idc)      # numero de filas = numero de elementos
idc$departamento  # el listado completo de la poblacion
```

Busquemos nuestro propio departamento, Norte de Santander, dentro de la población completa:

```r
idc[idc$departamento == "Norte de Santander", ]
```

## 3. De la población a la muestra

Ahora simulemos que, en lugar de tener el índice completo de los 33 entes territoriales, solo tuviéramos información de una **muestra** de 8 departamentos, seleccionados al azar.

Usamos `set.seed()` para que el sorteo aleatorio sea *reproducible*: si todos corremos el mismo código con la misma semilla, obtenemos exactamente la misma muestra.

```r
set.seed(2026)
posiciones_muestra <- sample(nrow(idc), 8)
muestra <- idc[posiciones_muestra, c("departamento", "region", "idc_2026")]
muestra
```

Comparemos el promedio del IDC 2026 calculado con la **muestra** frente al promedio calculado con la **población completa**. No tienen por qué ser exactamente iguales — esa diferencia es precisamente lo que en clases futuras llamaremos *error de muestreo*.

```r
mean(muestra$idc_2026)   # promedio en la muestra (8 departamentos)
mean(idc$idc_2026)       # promedio en la poblacion completa (33 entidades)
```

## 4. Identificación y clasificación de variables

Cada columna de `idc` es una variable. Vamos a identificarlas y clasificarlas según lo visto en la Unidad 1:

- **Cualitativas (nominales):** no tienen un orden natural. Ejemplo: `departamento`, `region`.
- **Cuantitativas continuas:** resultan de una medición y pueden tomar cualquier valor dentro de un rango. Ejemplo: `idc_2025`, `idc_2026`, y todos los puntajes por pilar (`instituciones_2025`, `salud_2026`, etc.).
- **Cuantitativas discretas / de posición:** `rank_2025`, `rank_2026` y `cambio_pos` son números enteros, pero representan una **posición** dentro de un ranking, no una magnitud medible directamente. Esto las acerca más a una variable ordinal que a una cuantitativa "pura" — es un buen ejemplo de que no toda variable numérica se comporta igual.

R nos puede confirmar cómo interpreta cada columna con `class()`:

```r
sapply(idc, class)
```

`unique()` nos muestra las categorías distintas de una variable cualitativa — útil para confirmar que, en efecto, no tiene un orden natural.

```r
unique(idc$region)
```

Construyamos una pequeña tabla de clasificación para algunas variables del conjunto de datos, como resumen de lo discutido en clase:

```r
clasificacion <- data.frame(
  variable = c("departamento", "region", "idc_2026", "instituciones_2026", "rank_2026", "cambio_pos"),
  tipo     = c("Cualitativa nominal", "Cualitativa nominal", "Cuantitativa continua",
               "Cuantitativa continua", "Cuantitativa discreta (posicional/ordinal)",
               "Cuantitativa discreta (posicional/ordinal)")
)
clasificacion
```
